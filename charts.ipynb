```python
import pandas as pd
import seaborn as sns
from scipy import stats

df = pd.read_excel("data/simplified.xlsx")


df.head()
```

```python
sns.violinplot(x="Modality", y="Knowledge acquisition", data=df,palette="pastel")
sns.swarmplot(x="Modality", y="Knowledge acquisition",color="darkred",size=7,data=df)
# sns.boxplot(x="Modality", y="Knowledge acquisition", data=df, showmeans=False,
            # showcaps=False, showbox=False, showfliers=False, medianprops={"color": "grey", "linewidth": 2})
```

```python
kruskal_result = stats.kruskal(df[df["Modality"] == "Immersive VR"]["Knowledge acquisition"],
                         df[df["Modality"] == "Desktop VR"]["Knowledge acquisition"],
                         df[df["Modality"] == "In presence"]["Knowledge acquisition"])
print("Kruskal-Wallis H-test result:", kruskal_result)
```

```python
sns.violinplot(x="Modality", y="Spatial awareness", data=df, palette="pastel")
sns.swarmplot(x="Modality", y="Spatial awareness",color="darkred",size=7,data=df)
sns.boxplot(x="Modality", y="Spatial awareness", data=df, showmeans=False,
            showcaps=False, showbox=False, showfliers=False, medianprops={"color": "grey", "linewidth": 2})
```

```python
kruskal_result = stats.kruskal(df[df["Modality"] == "Immersive VR"]["Spatial awareness"],
                         df[df["Modality"] == "Desktop VR"]["Spatial awareness"],
                         df[df["Modality"] == "In presence"]["Spatial awareness"])
print("Kruskal-Wallis H-test result:", kruskal_result)

import scikit_posthocs as sp
print("Significant difference found! Running Dunn's post-hoc test:")
posthoc_results = sp.posthoc_dunn(df, val_col='Spatial awareness', group_col='Modality', p_adjust='bonferroni')
print(posthoc_results)
```
