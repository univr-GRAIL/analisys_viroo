```python
import pandas as pd
import seaborn as sns
from scipy import stats
import matplotlib.pyplot as plt

plt.rcParams.update({
    "axes.labelsize": 16,   # x/y labels
    "xtick.labelsize": 14,  # x ticks
    "ytick.labelsize": 14,  # y ticks
    "legend.fontsize": 14,
    "axes.titlesize": 18
})
df = pd.read_excel("data/simplified.xlsx")


df.head()
df2=df[df["Modality"].isin(["Immersive VR", "Desktop VR"])]
```

```python
ax=sns.violinplot(x="Modality", y="Knowledge acquisition", data=df,palette="pastel")
sns.swarmplot(x="Modality", y="Knowledge acquisition",color="darkred",size=7,data=df)
ax.set_ylim(0, 1)
# sns.boxplot(x="Modality", y="Knowledge acquisition", data=df, showmeans=False,
            # showcaps=False, showbox=False, showfliers=False, medianprops={"color": "grey", "linewidth": 2})
```

```python
kruskal_result = stats.kruskal(df[df["Modality"] == "Immersive VR"]["Knowledge acquisition"],
                         df[df["Modality"] == "Desktop VR"]["Knowledge acquisition"],
                         df[df["Modality"] == "In presence"]["Knowledge acquisition"])
print("Kruskal-Wallis H-test result:", kruskal_result)
```

```python
import scikit_posthocs as sp
df.columns = df.columns.str.strip()
df["Modality"] = df["Modality"].astype(str).str.strip()
dunn = sp.posthoc_dunn(
    df,
    val_col="Knowledge acquisition",
    group_col="Modality",
    p_adjust="holm"  # or 'bonferroni', 'fdr_bh'
)
print("\nDunn's post-hoc test:")
print(dunn)
```

```python
anova = stats.f_oneway(df[df["Modality"] == "Immersive VR"]["Knowledge acquisition"],
                         df[df["Modality"] == "Desktop VR"]["Knowledge acquisition"],  
                            df[df["Modality"] == "In presence"]["Knowledge acquisition"]) 
print("anova_oneway result:", anova )
```
